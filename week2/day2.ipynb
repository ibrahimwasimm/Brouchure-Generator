```python
def stream_gpt(prompt):
    messages = [{"role":'system',"content":system_prompt}, 
    {"role":"user","content":prompt}]
    stream = ollama.chat.completions.create(model='minimax-m2.5:cloud', messages=messages, stream=True)
    response_chunk = ""
    for chunk in stream:
        response_chunk += chunk.choices[0].delta.content or ""
        yield response_chunk
```

```python
message_input = gr.Textbox(label="Your message:", info="Enter a message for GPT-4.1-mini", lines=7)
message_output = gr.Markdown(label="Response:")

view = gr.Interface(
    fn=stream_gpt,
    title="GPT", 
    inputs=[message_input], 
    outputs=[message_output], 
    examples=[
        "Explain the Transformer architecture to a layperson",
        "Explain the Transformer architecture to an aspiring AI engineer",
        ], 
    flagging_mode="never"
    )
view.launch()
```

```python
def stream_gemini(prompt):
    messages = [{"role":'system',"content":system_prompt}, 
    {"role":"user","content":prompt}]
    stream = gemini.chat.completions.create(model='gemini-2.5-flash-lite', messages=messages, stream=True)
    response_chunk = ""
    for chunk in stream:
        response_chunk += chunk.choices[0].delta.content or ""
        yield response_chunk
```

```python
def stream_model(prompt, model):
    if model=="ollama":
        result = stream_gpt(prompt)
    elif model=="gemini":
        result = stream_gemini(prompt)
    else:
        raise ValueError("Unknown model")
    yield from result
```

```python
message_input = gr.Textbox(label="Your message:", info="Enter a message for GPT-4.1-mini", lines=7)
model_selector = gr.Dropdown(["ollama", "gemini"], label="Select model", value="gemini")
message_output = gr.Markdown(label="Response:",)

view = gr.Interface(
    fn=stream_model,
    title="GPT", 
    inputs=[message_input, model_selector],  
    outputs=[message_output], 
    examples=[
        ["Explain the Transformer architecture to a layperson", "gemini"],
        ["Explain the Transformer architecture to an aspiring AI engineer", "ollama"],
        ], 
    flagging_mode="never"
    )
view.launch()
```
